```python
# Create a new schema to save the code descriptions
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.code_descriptions")
code_file_path = "/Volumes/workspace/default/uc_fund_data_2025_bronze/Code_Details/"
files = dbutils.fs.ls(code_file_path)

for file in files:
  file_name = file.name.replace(".csv", "")
  
  df = spark.read.option("header", True).option("delimiter", ",").option("inferSchema", True).csv(code_file_path + file.name)
  df = df.dropna(how='any')
  
  df.write.mode("overwrite").saveAsTable(f"workspace.code_descriptions.{file_name}")
  
```
